Q1: Student Management System (List-Based)

Mapped CO: CO2

Assignment Question:

Develop a Python program using lists to store student records and perform operations such as add, update, delete, and search.

```python


students = []

while True:
    print("\n1. Add Student")
    print("2. Update Student")
    print("3. Delete Student")
    print("4. Search Student")
    print("5. Display All")
    print("6. Exit")

    choice = input("Enter choice: ")

    if choice == "1":
        name = input("Enter student name: ")
        students.append(name)
        print("Student Added Successfully.")

    elif choice == "2":
        old = input("Enter name to update: ")
        if old in students:
            new = input("Enter new name: ")
            index = students.index(old)
            students[index] = new
            print("Student Updated.")
        else:
            print("Student not found.")

    elif choice == "3":
        name = input("Enter name to delete: ")
        if name in students:
            students.remove(name)
            print("Student Deleted.")
        else:
            print("Student not found.")

    elif choice == "4":
        name = input("Enter name to search: ")
        if name in students:
            print("Student Found.")
        else:
            print("Student not found.")

    elif choice == "5":
        print("Student List:", students)

    elif choice == "6":
        break

    else:
        print("Invalid Choice")
```

Q4: Document Organizer (Dictionary-Based)

Mapped CO: CO2

Assignment Question:

Develop a Python application using dictionaries to store document ID and document name and perform update, delete, and search operations.

```python
documents = {}

while True:
    print("\n1. Add Document")
    print("2. Update Document")
    print("3. Delete Document")
    print("4. Search Document")
    print("5. Display All")
    print("6. Exit")

    choice = input("Enter choice: ")

    if choice == "1":
        doc_id = input("Enter Document ID: ")
        doc_name = input("Enter Document Name: ")
        documents[doc_id] = doc_name
        print("Document Added.")

    elif choice == "2":
        doc_id = input("Enter Document ID to update: ")
        if doc_id in documents:
            documents[doc_id] = input("Enter new Document Name: ")
            print("Document Updated.")
        else:
            print("Document not found.")

    elif choice == "3":
        doc_id = input("Enter Document ID to delete: ")
        if doc_id in documents:
            del documents[doc_id]
            print("Document Deleted.")
        else:
            print("Document not found.")

    elif choice == "4":
        doc_id = input("Enter Document ID to search: ")
        if doc_id in documents:
            print("Document Name:", documents[doc_id])
        else:
            print("Document not found.")

    elif choice == "5":
        print("All Documents:", documents)
        

    elif choice == "6":
        break

    else:
        print("Invalid Choice")
```

Q5: Calculator Application (Data Structure-Based)

Mapped CO: CO2

Assignment Question:

Design a calculator program that stores calculation results and operation history using lists and dictionaries.

```python
history = []
results = {}

while True:
    print("\n1. Add")
    print("2. Subtract")
    print("3. Multiply")
    print("4. Divide")
    print("5. Show History")
    print("6. Exit")

    choice = input("Enter choice: ")

    if choice in ["1", "2", "3", "4"]:
        a = float(input("Enter first number: "))
        b = float(input("Enter second number: "))

        if choice == "1":
            result = a + b
            operation = "Addition"
        elif choice == "2":
            result = a - b
            operation = "Subtraction"
        elif choice == "3":
            result = a * b
            operation = "Multiplication"
        elif choice == "4":
            if b != 0:
                result = a / b
                operation = "Division"
            else:
                print("Division by zero not allowed.")
                continue

        history.append(f"{operation}: {a}, {b} = {result}")
        results[len(history)] = result

        print("Result:", result)

    elif choice == "5":
        print("\nCalculation History:")
        for item in history:
            print(item)

    elif choice == "6":
        break

    else:
        print("Invalid Choice")
```

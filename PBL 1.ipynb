2	Calculate equivalent resistance of resistors connected in series and parallel

```python
# Equivalent Resistance

R1 = float(input("Enter first resistance (Ohms): "))
R2 = float(input("Enter second resistance (Ohms): "))

# Series Resistance
R_series = R1 + R2

# Parallel Resistance
R_parallel = (R1 * R2) / (R1 + R2)

print("Equivalent Resistance in Series:", R_series, "Ohms")
print("Equivalent Resistance in Parallel:", R_parallel, "Ohms")
```

3	Calculate voltage using Ohm’s Law

```python
# Ohm's Law

I = float(input("Enter Current (Amperes): "))
R = float(input("Enter Resistance (Ohms): "))

V = I * R

print("Voltage =", V, "Volts")
```

4	Check whether the entered frequency is audio or radio frequency

```python
freq = float(input("Enter frequency in Hz: "))

if 20 <= freq <= 20000:
    print("It is an Audio Frequency.")
elif freq > 20000:
    print("It is a Radio Frequency.")
else:
    print("Below human hearing range.")
```

5	Display various radio frequency bands using if–elif ladder

```python
freq = float(input("Enter frequency in MHz: "))

if 0.3 <= freq < 3:
    print("MF - Medium Frequency")
elif 3 <= freq < 30:
    print("HF - High Frequency")
elif 30 <= freq < 300:
    print("VHF - Very High Frequency")
elif 300 <= freq < 3000:
    print("UHF - Ultra High Frequency")
else:
    print("Frequency out of common RF band range.")
```

6	Display resistor color code using match-case

```python
color = input("Enter resistor color: ").lower()

match color:
    case "black":
        print("Value: 0")
    case "brown":
        print("Value: 1")
    case "red":
        print("Value: 2")
    case "orange":
        print("Value: 3")
    case "yellow":
        print("Value: 4")
    case "green":
        print("Value: 5")
    case "blue":
        print("Value: 6")
    case "violet":
        print("Value: 7")
    case "grey":
        print("Value: 8")
    case "white":
        print("Value: 9")
    case _:
        print("Invalid Color")
```

1: Creating Sets

```python

my_set = {1, 2, 3, 4, 5}
print("Set:", my_set)

my_set2 = set([1, 2, 3, 4, 5])
print("Set using set():", my_set2)

empty_set = set()
print("Empty Set:", empty_set)
```

2: Accessing Elements (Using Loop)

```python
my_set = {10, 20, 30, 40}

for item in my_set:
    print(item)
```

3: Adding Elements to a Set

```python
my_set = {1, 2, 3}
print("Original Set:", my_set)

my_set.add(4)
print("After adding 4:", my_set)

my_set.update([5, 6])
print("After adding 5 and 6:", my_set)
```

4: Removing Elements from a Set

```python
my_set = {1, 2, 3, 4, 5}
print("Original Set:", my_set)

my_set.remove(4)
print("After removing 4:", my_set)

my_set.discard(3)
print("After discarding 3:", my_set)

popped = my_set.pop()
print("Popped Element:", popped)
print("Set after pop:", my_set)
```

5: Membership Testing

```python
my_set = {1, 2, 3}

print(1 in my_set)
print(4 in my_set)
```

    6: Set Operations

```python
set1 = {1, 2, 3}
set2 = {3, 4, 5}

print("Union:", set1.union(set2))
print("Intersection:", set1.intersection(set2))
print("Difference:", set1.difference(set2))
print("Symmetric Difference:", set1.symmetric_difference(set2))
```

Shopping Cart System

```python
cart = set()

while True:
    print("\n--- Shopping Cart System ---")
    print("1. Add Product ID")
    print("2. Display Cart")
    print("3. Total Products")
    print("4. Highest & Lowest Product ID")
    print("5. Search Product")
    print("6. Remove Product")
    print("7. Premium Products")
    print("8. Union & Intersection with Another Cart")
    print("9. Display Sorted Cart")
    print("10. Exit")

    choice = int(input("Enter choice: "))

    if choice == 1:
        pid = int(input("Enter product ID: "))
        cart.add(pid)

    elif choice == 2:
        print("Cart Items:", cart)

    elif choice == 3:
        print("Total unique products:", len(cart))

    elif choice == 4:
        print("Highest ID:", max(cart))
        print("Lowest ID:", min(cart))

    elif choice == 5:
        pid = int(input("Enter product ID to search: "))
        print("Found" if pid in cart else "Not Found")

    elif choice == 6:
        pid = int(input("Enter product ID to remove: "))
        cart.discard(pid)
        print("Product removed")

    elif choice == 7:
        limit = int(input("Enter premium limit: "))
        print([p for p in cart if p >= limit])

    elif choice == 8:
        other_cart = {101, 102, 103}
        print("Union:", cart.union(other_cart))
        print("Intersection:", cart.intersection(other_cart))

    elif choice == 9:
        print("Sorted Cart:", sorted(cart))

    elif choice == 10:
        break
```

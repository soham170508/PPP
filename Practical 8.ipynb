```python
def half_adder(a, b):
    sum = a ^ b
    carry = a & b
    return sum, carry

def full_adder(a, b, c):
    sum = a ^ b ^ c
    carry = (a & b) | (b & c) | (a & c)
    return sum, carry

a = int(input("Enter first bit: "))
b = int(input("Enter second bit: "))
c = int(input("Enter carry bit: "))

hs, hc = half_adder(a, b)
fs, fc = full_adder(a, b, c)

print("Half Adder Sum:", hs, "Carry:", hc)
print("Full Adder Sum:", fs, "Carry:", fc)
```

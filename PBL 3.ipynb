2. User-Defined Functions

Develop a Python program using user-defined functions to implement Half Adder and Full Adder logic.


```python
# Half Adder
def half_adder(a, b):
    sum_bit = a ^ b
    carry = a & b
    return sum_bit, carry

# Full Adder
def full_adder(a, b, cin):
    sum_bit = a ^ b ^ cin
    carry = (a & b) | (b & cin) | (a & cin)
    return sum_bit, carry
    
print("Half Adder (1,1):", half_adder(1,1))
print("Full Adder (1,1,1):", full_adder(1,1,1))
```

3. Classes and Objects

Write a Python program to define a Student class with appropriate data members and member functions.
Create objects of the class and display student details.


```python
class Student:
    def __init__(self, roll, name, marks):
        self.roll = roll
        self.name = name
        self.marks = marks
    
    def display(self):
        print("Roll No:", self.roll)
        print("Name:", self.name)
        print("Marks:", self.marks)
        print("--------------------")

s1 = Student(1, "Soham", 85)
s2 = Student(2, "Amit", 78)

s1.display()
s2.display()
```

7. User-Defined Package

Create and use a user-defined Python package for organizing multiple related modules.

```python
import os

# Create package folder
os.makedirs("mypackage", exist_ok=True)

# __init__.py
with open("mypackage/__init__.py", "w") as f:
    pass

# Module 1
with open("mypackage/module1.py", "w") as f:
    f.write("""
def greet(name):
    return f"Hello {name}"
""")

# Module 2
with open("mypackage/module2.py", "w") as f:
    f.write("""
def square(x):
    return x * x
""")

from mypackage import module1, module2

print(module1.greet("Soham"))
print("Square:", module2.square(5))
```

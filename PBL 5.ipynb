1. Stack Implementation

Write a Python program to implement a Stack using array with PUSH and POP operations for managing records.



```python
# Stack Implementation

stack = []

def push(data):
    stack.append(data)
    print(data, "pushed into stack")

def pop():
    if stack:
        print(stack.pop(), "popped from stack")
    else:
        print("Stack is Empty")

def display():
    print("Stack:", stack)


# Testing
push("Record1")
push("Record2")
display()
pop()
display()
```

2. Queue Implementation

Develop a Python program to implement a Queue using array with ENQUEUE and DEQUEUE operations for data processing.




```python
# Queue Implementation

queue = []

def enqueue(data):
    queue.append(data)
    print(data, "added to queue")

def dequeue():
    if queue:
        print(queue.pop(0), "removed from queue")
    else:
        print("Queue is Empty")

def display():
    print("Queue:", queue)


# Testing
enqueue("Task1")
enqueue("Task2")
display()
dequeue()
display()
```

```python
# Node Class
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

# Linked List Class
class LinkedList:
    def __init__(self):
        self.head = None

    def insert(self, data):
        new_node = Node(data)
        if self.head is None:
            self.head = new_node
        else:
            temp = self.head
            while temp.next:
                temp = temp.next
            temp.next = new_node

    def delete(self, key):
        temp = self.head

        if temp and temp.data == key:
            self.head = temp.next
            temp = None
            return

        prev = None
        while temp and temp.data != key:
            prev = temp
            temp = temp.next

        if temp is None:
            print("Element not found")
            return

        prev.next = temp.next
        temp = None

    def display(self):
        temp = self.head
        while temp:
            print(temp.data, end=" -> ")
            temp = temp.next
        print("None")


# Testing
ll = LinkedList()
ll.insert("A")
ll.insert("B")
ll.insert("C")
ll.display()
ll.delete("B")
ll.display()
```

5. Data Structure Selection

Analyze and justify the selection of appropriate linear data structures for different operations used in the mini project.

1. Stack
Purpose Used:

Managing undo operations / last inserted record removal.

Reason for Selection:

Stack follows LIFO (Last In First Out) principle.

Suitable when the most recent data needs to be accessed or removed first.

Simple implementation using Python list.

Justification:

When record rollback or recent activity tracking is required, stack is the most efficient linear data structure.

2. Queue
Purpose Used:

Task scheduling / request processing.

Reason for Selection:

Queue follows FIFO (First In First Out) principle.

Suitable for processing data in the order of arrival.

Ensures fairness in handling requests.

Justification:

For real-life scenarios like lab request processing or student service handling, queue ensures systematic processing.

3. Singly Linked List
Purpose Used:

Dynamic insertion and deletion of records.

Reason for Selection:

Does not require contiguous memory allocation.

Efficient for frequent insertions and deletions.

Flexible structure compared to arrays.

Justification:

Linked list is suitable where dynamic memory management is required and data size may vary.

4. List and Dictionary (Mini Project – Student Management System)
Purpose Used:

Storing structured student records.

Reason for Selection:

List allows storing multiple student records.

Dictionary allows storing data in key–value format (Roll No, Name, Marks).

Easy data access, update, and traversal.

Justification:

Combination of list and dictionary provides both flexibility and structured storage, making it ideal for managing student data.

Conclusion

The selection of appropriate linear data structures was based on operational requirements:

Stack for LIFO operations

Queue for FIFO processing

Linked List for dynamic memory management

List & Dictionary for structured record handling

```python
n = 3
identity_matrix = []

for i in range(n):
    row = []
    for j in range(n):
        if i == j:
            row.append(1)
        else:
            row.append(0)
    identity_matrix.append(row)

for row in identity_matrix:
    print(row)
```

```python
matrix = []
value = 2

for i in range(4):
    row = []
    for j in range(4):
        row.append(value)
        value += 1
    matrix.append(row)

for row in matrix:
    print(row)
```

```python
# Creating a user-defined package using ModuleType (Jupyter safe)

import types

mypackage = types.ModuleType("mypackage")
ops = types.ModuleType("ops")

def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

ops.add = add
ops.subtract = subtract
mypackage.ops = ops

print("Addition:", mypackage.ops.add(10, 5))
print("Subtraction:", mypackage.ops.subtract(10, 5))

```

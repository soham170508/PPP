Q1: Student Management System (Function-Based)

Mapped CO: CO3

Assignment Question:

Develop a Python program using user-defined functions to add, modify, and display student records.

```python
# Student Management using Functions

students = []

def add_student(name):
    students.append(name)
    print("Student Added Successfully.")

def modify_student(old_name, new_name):
    if old_name in students:
        index = students.index(old_name)
        students[index] = new_name
        print("Student Modified Successfully.")
    else:
        print("Student Not Found.")

def display_students():
    print("Student Records:")
    for student in students:
        print(student)

# Main Program
while True:
    print("\n1. Add Student")
    print("2. Modify Student")
    print("3. Display Students")
    print("4. Exit")

    choice = input("Enter choice: ")

    if choice == "1":
        name = input("Enter student name: ")
        add_student(name)

    elif choice == "2":
        old = input("Enter existing name: ")
        new = input("Enter new name: ")
        modify_student(old, new)

    elif choice == "3":
        display_students()

    elif choice == "4":
        break
```

Q3: Electronics Tools Management System (Package-Based)

Mapped CO: CO3

Assignment Question:

Develop a Python package consisting of multiple modules to manage electronic tools inventory and pricing.

```python
# Inventory Module
tools = {}

def add_tool(name, quantity):
    tools[name] = quantity

def display_tools():
    print("Inventory:")
    for tool, qty in tools.items():
        print(tool, ":", qty)

# Pricing Module
prices = {}

def set_price(name, price):
    prices[name] = price

def display_prices():
    print("Pricing:")
    for tool, price in prices.items():
        print(tool, "Price:", price)
        
add_tool("Oscilloscope", 5)
set_price("Oscilloscope", 25000)

display_tools()
display_prices()
```

Q5: Calculator Application (Class-Based)

Mapped CO: CO3

Assignment Question:

Implement a calculator using classes and objects with constructors and member functions.

```python
class Calculator:

    def __init__(self, a, b):
        self.a = a
        self.b = b

    def add(self):
        return self.a + self.b

    def subtract(self):
        return self.a - self.b

    def multiply(self):
        return self.a * self.b

    def divide(self):
        if self.b != 0:
            return self.a / self.b
        else:
            return "Division by zero not allowed"
calc = Calculator(10, 5)

print("Addition:", calc.add())
print("Subtraction:", calc.subtract())
print("Multiplication:", calc.multiply())
print("Division:", calc.divide())
```

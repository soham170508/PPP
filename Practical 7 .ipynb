1: Creating a Dictionary

```python
student = {
    "name": "John",
    "age": 30,
    "city": "New York"
}

print(student)
```

2: Accessing Dictionary Values

```python
print("Name:", student["name"])
print("Age:", student["age"])
print("City:", student["city"])
```

3: Updating Dictionary Values

```python
student["age"] = 35
print(student)
```

```python
student["gender"] = "Male"
print(student)
```

4: Deleting Dictionary Elements

```python

del student["city"]
print(student)

student.clear()
print(student)
```

Bank Account System

```python
accounts = {}

while True:
    print("\n--- Bank Account Management System ---")
    print("1. Create Account")
    print("2. Display Accounts")
    print("3. Total & Average Balance")
    print("4. Highest & Lowest Balance")
    print("5. High Value Accounts")
    print("6. Search Account")
    print("7. Deposit / Withdraw")
    print("8. Remove Account")
    print("9. Display Keys & Values")
    print("10. Sort Accounts")
    print("11. Exit")

    choice = int(input("Enter choice: "))

    if choice == 1:
        acc = int(input("Enter account number: "))
        bal = float(input("Enter balance: "))
        accounts[acc] = bal

    elif choice == 2:
        print("Accounts:", accounts)

    elif choice == 3:
        total = sum(accounts.values())
        print("Total Balance:", total)
        print("Average Balance:", total / len(accounts))

    elif choice == 4:
        print("Highest Balance:", max(accounts.values()))
        print("Lowest Balance:", min(accounts.values()))

    elif choice == 5:
        limit = float(input("Enter limit: "))
        for acc, bal in accounts.items():
            if bal >= limit:
                print(acc, ":", bal)

    elif choice == 6:
        acc = int(input("Enter account number: "))
        print("Found" if acc in accounts else "Not Found")

    elif choice == 7:
        acc = int(input("Enter account number: "))
        if acc in accounts:
            amt = float(input("Enter amount (+deposit / -withdraw): "))
            accounts[acc] += amt
            print("Updated Balance:", accounts[acc])

    elif choice == 8:
        acc = int(input("Enter account number to delete: "))
        if acc in accounts:
            del accounts[acc]
            print("Account removed")

    elif choice == 9:
        print("Account Numbers:", accounts.keys())
        print("Balances:", accounts.values())

    elif choice == 10:
        for acc in sorted(accounts):
            print(acc, ":", accounts[acc])

    elif choice == 11:
        break
```

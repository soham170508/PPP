2. Accessing Elements

```python
my_list = [10, 20, 30, 40, 50]
print(my_list[0])    
print(my_list[-1])   
```

3. Slicing Lists

```python
my_list = [10, 20, 30, 40, 50]
print(my_list[1:4])  
print(my_list[:3])  
print(my_list[2:])   
```

4. Modifying Lists

```python
my_list = [10, 20, 30, 40, 50]

my_list[1] = 25        
my_list.append(60)    
my_list.insert(2, 35) 
my_list.remove(40)    
del my_list[0]  
print(my_list)
```

5. List Operations

```python
list1 = [1, 2, 3]
list2 = [4, 5, 6]

print(list1 + list2)
print(list1 * 3)
print(2 in list1)
```

Student Management System (Menu-Driven Program)

```python
marks = []

while True:
    print("\n--- Student Management System ---")
    print("1. Add Student Marks")
    print("2. Display Marks")
    print("3. Total and Average")
    print("4. Highest and Lowest Marks")
    print("5. Passed and Failed Students")
    print("6. Distinction Students")
    print("7. Remove Duplicates")
    print("8. Replace Failed with Grace Marks")
    print("9. Search Marks")
    print("10. Sort and Reverse")
    print("11. Remove Student Record")
    print("12. Find Student Position")
    print("13. Exit")

    choice = int(input("Enter choice: "))

    if choice == 1:
        n = int(input("How many students? "))
        for i in range(n):
            marks.append(int(input("Enter marks: ")))

    elif choice == 2:
        print("Marks:", marks)

    elif choice == 3:
        print("Total:", sum(marks))
        print("Average:", sum(marks)/len(marks))

    elif choice == 4:
        print("Highest:", max(marks))
        print("Lowest:", min(marks))

    elif choice == 5:
        print("Passed:", [m for m in marks if m >= 40])
        print("Failed:", [m for m in marks if m < 40])

    elif choice == 6:
        print("Distinction:", [m for m in marks if m >= 75])

    elif choice == 7:
        marks = list(set(marks))
        print("Duplicates removed")

    elif choice == 8:
        marks = [40 if m < 40 else m for m in marks]
        print("Grace marks applied")

    elif choice == 9:
        s = int(input("Enter marks to search: "))
        print("Found" if s in marks else "Not Found")

    elif choice == 10:
        marks.sort()
        print("Sorted:", marks)
        print("Reversed:", marks[::-1])

    elif choice == 11:
        r = int(input("Enter marks to remove: "))
        if r in marks:
            marks.remove(r)

    elif choice == 12:
        m = int(input("Enter marks: "))
        if m in marks:
            print("Position:", marks.index(m))

    elif choice == 13:
        break
```

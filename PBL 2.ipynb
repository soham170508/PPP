1. List Operations
Develop a Python program to create, access, update, and delete student records using a List data structure.

```python

students = ["Amit", "Soham", "Riya"]


print("First Student:", students[0])


students[1] = "Rahul"


students.append("Neha")


students.remove("Amit")

print("Updated Student List:", students)
```

2. Tuple Operations
Write a Python program to store fixed laboratory information using a Tuple and perform access and modification operations.

```python

lab_info = ("Electronics Lab", 101, "Analog Circuits")
print("Lab Name:", lab_info[0])
print("Room No:", lab_info[1])
lab_list = list(lab_info)
lab_list[1] = 102
lab_info = tuple(lab_list)

print("Modified Lab Info:", lab_info)
```

3. Set Operations
Develop a Python program to manage unique laboratory equipment records using a Set and perform add and remove operations.

```python

equipment = {"Oscilloscope", "Multimeter", "Power Supply"}


equipment.add("Function Generator")


equipment.remove("Power Supply")

print("Available Equipment:", equipment)
```

4. Dictionary Operations
Write a Python program to store and manage student details (Roll No, Name, Marks) using a Dictionary, including create, access, update, delete, and traversal operations.


```python

student = {
    "Roll No": 101,
    "Name": "Soham",
    "Marks": 85
}


print("Student Name:", student["Name"])


student["Marks"] = 90


student["Department"] = "ECE"


del student["Department"]


for key, value in student.items():
    print(key, ":", value)
```

5. Student Record Management
Design a Python program using appropriate data structures to maintain and display multiple student records.

```python

students = [
    {"Roll": 1, "Name": "Amit", "Marks": 78},
    {"Roll": 2, "Name": "Soham", "Marks": 85},
    {"Roll": 3, "Name": "Riya", "Marks": 92}
]

print("Student Records:")
for s in students:
    print(s)
```

6. Laboratory Record Handling
Develop a Python program to manage laboratory resources and availability using Python data structures.

```python

lab_resources = {
    "Oscilloscope": 5,
    "Multimeter": 10,
    "Power Supply": 4
}


lab_resources["Multimeter"] = 8


lab_resources["Function Generator"] = 3

print("Laboratory Resources:")
for item, qty in lab_resources.items():
    print(item, ":", qty)
```

7. Data Traversal
Write a Python program to traverse and display all stored student and laboratory records efficiently.



```python

students = [
    {"Roll": 1, "Name": "Amit", "Marks": 78},
    {"Roll": 2, "Name": "Soham", "Marks": 85}
]


lab_resources = {
    "Oscilloscope": 5,
    "Multimeter": 8
}

print("---- Student Records ----")
for student in students:
    for key, value in student.items():
        print(key, ":", value)
    print()

print("---- Laboratory Records ----")
for item, qty in lab_resources.items():
    print(item, ":", qty)
```

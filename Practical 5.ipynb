1.Creating Tuples

```python
empty_tuple = ()
single_element_tuple = (5,)
multiple_elements_tuple = (1, 2, 3, 'a', 'b', 'c')
nested_tuple = ((1, 2), ('a', 'b'))

print(empty_tuple)
print(single_element_tuple)
print(multiple_elements_tuple)
print(nested_tuple)
```

2.Accessing Tuple Elements

```python
my_tuple = ('apple', 'banana', 'cherry', 'date')

print("First element:", my_tuple[0])
print("Last element:", my_tuple[-1])
```

3: Slicing Tuples

```python
print(my_tuple[1:3])
print(my_tuple[:2])
print(my_tuple[2:])
```

5: Updating Tuple (Indirect Method)

```python
my_tuple = (1, 2, 3)

temp = list(my_tuple)
temp[0] = 10
my_tuple = tuple(temp)

print(my_tuple)
```

6: Deleting Tuple Elements (Indirect)

```python
my_tuple = (10, 20, 30, 40)

temp = list(my_tuple)
del temp[2]
my_tuple = tuple(temp)

print(my_tuple)
```

 7: Tuple Packing and Unpacking

```python
my_tuple = 10, 20, 30

a, b, c = my_tuple

print(a)
print(b)
print(c)
```

8: Tuple as Dictionary Keys

```python
my_dict = {
    ('Book1', 101): 'Available',
    ('Book2', 102): 'Issued'
}

print(my_dict[('Book1', 101)])
print(my_dict[('Book2', 102)])
```

Library Inventory System

```python
books = ()

while True:
    print("\n--- Library Inventory System ---")
    print("1. Add Book Values")
    print("2. Display Books")
    print("3. Total & Average")
    print("4. Highest & Lowest")
    print("5. Premium Books")
    print("6. Search Book")
    print("7. Count Book Value")
    print("8. Remove Duplicates")
    print("9. Reverse Tuple")
    print("10. Find Position")
    print("11. Exit")

    choice = int(input("Enter choice: "))

    if choice == 1:
        temp = list(books)
        n = int(input("How many books? "))
        for i in range(n):
            temp.append(int(input("Enter book value: ")))
        books = tuple(temp)

    elif choice == 2:
        print("Books:", books)

    elif choice == 3:
        print("Total Books:", len(books))
        print("Average Value:", sum(books)/len(books))

    elif choice == 4:
        print("Highest:", max(books))
        print("Lowest:", min(books))

    elif choice == 5:
        limit = int(input("Enter limit: "))
        print([b for b in books if b >= limit])

    elif choice == 6:
        v = int(input("Enter value to search: "))
        print("Found" if v in books else "Not Found")

    elif choice == 7:
        v = int(input("Enter value to count: "))
        print("Count:", books.count(v))

    elif choice == 8:
        books = tuple(set(books))
        print("Duplicates removed")

    elif choice == 9:
        print("Reversed:", books[::-1])

    elif choice == 10:
        v = int(input("Enter value: "))
        if v in books:
            print("Position:", books.index(v))

    elif choice == 11:
        break
```

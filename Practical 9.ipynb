```python
class Student:
    def get_data(self):
        self.roll = int(input("Enter Roll No: "))
        self.name = input("Enter Name: ")
        self.course = input("Enter Course: ")
        self.percentage = float(input("Enter Percentage: "))

    def display(self):
        print("\nStudent Details")
        print("Roll No:", self.roll)
        print("Name:", self.name)
        print("Course:", self.course)
        print("Percentage:", self.percentage)

s1 = Student()
s1.get_data()
s1.display()
```

```python
class Student:
    def __init__(self, roll, name, course, percentage):
        self.roll = roll
        self.name = name
        self.course = course
        self.percentage = percentage

    def display(self):
        print("\nStudent Details")
        print("Roll No:", self.roll)
        print("Name:", self.name)
        print("Course:", self.course)
        print("Percentage:", self.percentage)

s1 = Student(2, "Rahul", "Electronics", 78.6)
s1.display()
```

```python
class Student:
    def __init__(self, roll, name, course, percentage):
        self.roll = roll
        self.name = name
        self.course = course
        self.percentage = percentage

    def display(self):
        print(self.roll, self.name, self.course, self.percentage)

n = int(input("Enter number of students: "))
students = []

for i in range(n):
    print("\nEnter details of student", i + 1)
    roll = int(input("Roll No: "))
    name = input("Name: ")
    course = input("Course: ")
    percentage = float(input("Percentage: "))

    students.append(Student(roll, name, course, percentage))

print("\nStudent Records:")
for s in students:
    s.display()
```

```python
def linear_search(arr, key):
    for i in range(len(arr)):
        if arr[i] == key:
            return i
    return -1

# Input
arr = [10, 20, 30, 40, 50]
key = 30

result = linear_search(arr, key)

if result != -1:
    print("Element found at index", result)
else:
    print("Element not found")
```

```python

def binary_search(arr, key):
    low = 0
    high = len(arr) - 1
    
    while low <= high:
        mid = (low + high) // 2
        
        if arr[mid] == key:
            return mid
        elif arr[mid] < key:
            low = mid + 1
        else:
            high = mid - 1
            
    return -1

# Input (Array must be sorted)
arr = [10, 20, 30, 40, 50]
key = 40

result = binary_search(arr, key)

if result != -1:
    print("Element found at index", result)
else:
    print("Element not found")
```

```python

def search_char(string, key):
    for i in range(len(string)):
        if string[i] == key:
            return i
    return -1

text = "PYTHON"
char = "T"

result = search_char(text, char)

if result != -1:
    print("Character found at index", result)
else:
    print("Character not found")
```

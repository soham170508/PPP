```python


stack = []

def push(element):
    stack.append(element)
    print(element, "pushed into stack")

def pop():
    if len(stack) == 0:
        print("Stack is Empty")
    else:
        print(stack.pop(), "popped from stack")

def display():
    print("Stack Elements:", stack)

push(10)
push(20)
push(30)
display()
pop()
display()
```

```python
# Queue Implementation using List (Array)

queue = []

# ENQUEUE (Insert)
def enqueue(element):
    queue.append(element)
    print(element, "inserted into queue")

# DEQUEUE (Delete)
def dequeue():
    if len(queue) == 0:
        print("Queue is Empty")
    else:
        print(queue.pop(0), "removed from queue")

# UPDATE
def update(index, value):
    if 0 <= index < len(queue):
        queue[index] = value
        print("Queue updated")
    else:
        print("Invalid index")

# DISPLAY
def display():
    print("Queue Elements:", queue)


# Testing
enqueue(100)
enqueue(200)
enqueue(300)
display()
update(1, 250)
display()
dequeue()
display()
```

```python
data = []

# Stack Operations
def push(x):
    data.append(x)
    print(x, "pushed (Stack)")

def pop_stack():
    if data:
        print(data.pop(), "popped (Stack)")
    else:
        print("Stack Empty")

# Queue Operations
def enqueue(x):
    data.append(x)
    print(x, "enqueued (Queue)")

def dequeue():
    if data:
        print(data.pop(0), "dequeued (Queue)")
    else:
        print("Queue Empty")

def display():
    print("Current Data:", data)


# Testing Stack
push(10)
push(20)
display()
pop_stack()
display()

print("-----")

# Testing Queue
enqueue(100)
enqueue(200)
display()
dequeue()
display()
```
